# Task 2: Data Cleaning and Preprocessing - Cardiovascular Disease Dataset

Input: `cardio_train.csv`

Output: cleaned/preprocessed data ready for Machine Learning.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("cardio_train.csv", sep=";")
df.head()
```

## 1. Check Dataset

```python
print("Shape:", df.shape)
df.info()
print("\nMissing values:")
print(df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
```

## 2. Remove Duplicate Rows

```python
before = len(df)
df = df.drop_duplicates().copy()
after = len(df)

print("Rows before:", before)
print("Rows after :", after)
print("Duplicates removed:", before - after)
```

## 3. Check Outliers using Boxplots

```python
numeric_cols = ["age", "height", "weight", "ap_hi", "ap_lo"]

for col in numeric_cols:
    plt.figure(figsize=(7,4))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.show()
```

## 4. Convert Age from Days to Years

```python
df["age_years"] = (df["age"] / 365.25).round(1)

print(df[["age", "age_years"]].head())
```

## 5. Handle Implausible Blood Pressure Values

For this preprocessing step, keep physiologically implausible blood-pressure records out of the modeling data.

```python
df = df[
    (df["ap_hi"] >= 70) & (df["ap_hi"] <= 250) &
    (df["ap_lo"] >= 40) & (df["ap_lo"] <= 150) &
    (df["ap_hi"] >= df["ap_lo"])
].copy()

print("Shape after BP cleaning:", df.shape)
```

## 6. Handle Height and Weight Outliers

```python
df = df[
    (df["height"] >= 100) & (df["height"] <= 250) &
    (df["weight"] >= 30) & (df["weight"] <= 200)
].copy()

print("Shape after height/weight cleaning:", df.shape)
```

## 7. Check Missing Values Again

```python
print(df.isnull().sum())
```

## 8. Select Features and Target

```python
X = df.drop(columns=["cardio"])
y = df["cardio"]

# ID is only an identifier and is not useful as a predictive feature
if "id" in X.columns:
    X = X.drop(columns=["id"])

print("Feature columns:")
print(X.columns.tolist())

print("\nTarget:")
print(y.name)
```

## 9. Feature Scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled = pd.DataFrame(
    X_scaled,
    columns=X.columns,
    index=X.index
)

X_scaled.head()
```

## 10. Final Preprocessed Dataset

```python
preprocessed_df = X_scaled.copy()
preprocessed_df["cardio"] = y.values

print("Final shape:", preprocessed_df.shape)
print("\nMissing values:")
print(preprocessed_df.isnull().sum())
print("\nPreview:")
preprocessed_df.head()
```

## 11. Final Checklist

- Loaded the cardiovascular disease dataset
- Checked shape, columns and data types
- Checked missing values
- Removed duplicate rows
- Checked numeric outliers
- Cleaned implausible blood-pressure values
- Cleaned height and weight values
- Converted age from days to years
- Removed `id` from model features
- Scaled numerical features
- Created `cardio_preprocessed.csv`
