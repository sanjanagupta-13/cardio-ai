# Cardio Dataset — Task 1: EDA

This notebook follows the **same 20 questions** from your original Task 1. The incomplete/incorrect parts have been completed and cleaned up.

Dataset: `cardio_train.csv`

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("cardio_train.csv", sep=';')
df.head(20)
```

**1. What is the shape of the dataset?**

```python
# shape means (rows, columns)
print("Shape of dataset:", df.shape)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

**2. What are the names of all the columns?**

```python
print("Column names:")
print(df.columns.tolist())
```

**3. What are the data types of each column?**

```python
print("Data types of each column:")
print(df.dtypes)
```

**4. Are there any missing (null) values in the dataset?**

```python
missing_values = df.isnull().sum()
print("Missing values in each column:")
print(missing_values)

print("\nTotal missing values:", missing_values.sum())
```

**5. Are there any duplicate records?**

```python
duplicates = df.duplicated().sum()
print("Number of duplicate records:", duplicates)
```

**6. What are the statistical summaries (mean, min, max, etc.) of the numerical features?**

```python
print("Statistical summary:")
df.describe()
```

**7. What is the distribution of patients' ages?**

```python
# Age is stored in days, so convert it approximately to years for EDA
age_years = df["age"] / 365.25

plt.figure(figsize=(8,5))
plt.hist(age_years, bins=20, edgecolor='black')
plt.title("Age Distribution of Patients")
plt.xlabel("Age (Years)")
plt.ylabel("Number of Patients")
plt.show()

print("Minimum age:", round(age_years.min(), 2), "years")
print("Maximum age:", round(age_years.max(), 2), "years")
print("Average age:", round(age_years.mean(), 2), "years")
```

**8. What is the distribution of patients' heights and weights?**

```python
# Height distribution
plt.figure(figsize=(8,5))
plt.hist(df["height"], bins=20, edgecolor='black')
plt.title("Height Distribution of Patients")
plt.xlabel("Height (cm)")
plt.ylabel("Number of Patients")
plt.show()

# Weight distribution
plt.figure(figsize=(8,5))
plt.hist(df["weight"], bins=20, edgecolor='black')
plt.title("Weight Distribution of Patients")
plt.xlabel("Weight (kg)")
plt.ylabel("Number of Patients")
plt.show()

print("Average height:", round(df["height"].mean(), 2), "cm")
print("Average weight:", round(df["weight"].mean(), 2), "kg")
```

**9. What is the distribution of systolic blood pressure (ap_hi)?**

```python
plt.figure(figsize=(8,5))
plt.hist(df["ap_hi"], bins=50, edgecolor='black')
plt.title("Distribution of Systolic Blood Pressure (ap_hi)")
plt.xlabel("Systolic Blood Pressure (mmHg)")
plt.ylabel("Number of Patients")
plt.show()

print("Minimum ap_hi:", df["ap_hi"].min())
print("Maximum ap_hi:", df["ap_hi"].max())
print("Average ap_hi:", round(df["ap_hi"].mean(), 2))
```

**10. What is the distribution of diastolic blood pressure (ap_lo)?**

```python
plt.figure(figsize=(8,5))
plt.hist(df["ap_lo"], bins=50, edgecolor='black')
plt.title("Distribution of Diastolic Blood Pressure (ap_lo)")
plt.xlabel("Diastolic Blood Pressure (mmHg)")
plt.ylabel("Number of Patients")
plt.show()

print("Minimum ap_lo:", df["ap_lo"].min())
print("Maximum ap_lo:", df["ap_lo"].max())
print("Average ap_lo:", round(df["ap_lo"].mean(), 2))
```

**11. How many patients have cardiovascular disease (cardio = 1)?**

```python
cardio_yes = (df["cardio"] == 1).sum()
print("Patients with cardiovascular disease (cardio = 1):", cardio_yes)
```

**12. How many patients do not have cardiovascular disease (cardio = 0)?**

```python
cardio_no = (df["cardio"] == 0).sum()
print("Patients without cardiovascular disease (cardio = 0):", cardio_no)
```

**13. Is the target variable (cardio) balanced or imbalanced?**

```python
cardio_counts = df["cardio"].value_counts().sort_index()
cardio_percent = df["cardio"].value_counts(normalize=True).sort_index() * 100

print("Counts:")
print(cardio_counts)

print("\nPercentage:")
print(cardio_percent.round(2))

difference = abs(cardio_percent[0] - cardio_percent[1])

if difference < 5:
    print("\nConclusion: The target variable is approximately balanced.")
else:
    print("\nConclusion: The target variable is imbalanced.")
```

**14. What is the distribution of gender in the dataset?**

```python
gender_counts = df["gender"].value_counts().sort_index()
print("Gender distribution:")
print(gender_counts)

plt.figure(figsize=(7,5))
plt.bar(gender_counts.index.astype(str), gender_counts.values)
plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Patients")
plt.show()
```

**15. How many patients have high cholesterol levels?**

```python
# cholesterol: 1 = normal, 2 = above normal, 3 = well above normal
cholesterol_counts = df["cholesterol"].value_counts().sort_index()

print("Cholesterol distribution:")
print(cholesterol_counts)

high_cholesterol = (df["cholesterol"] > 1).sum()
print("\nPatients with high cholesterol (above normal or well above normal):", high_cholesterol)

plt.figure(figsize=(7,5))
plt.bar(cholesterol_counts.index.astype(str), cholesterol_counts.values)
plt.title("Cholesterol Level Distribution")
plt.xlabel("Cholesterol Level (1=Normal, 2=Above Normal, 3=Well Above Normal)")
plt.ylabel("Number of Patients")
plt.show()
```

**16. How many patients have high glucose levels?**

```python
# gluc: 1 = normal, 2 = above normal, 3 = well above normal
glucose_counts = df["gluc"].value_counts().sort_index()

print("Glucose distribution:")
print(glucose_counts)

high_glucose = (df["gluc"] > 1).sum()
print("\nPatients with high glucose (above normal or well above normal):", high_glucose)

plt.figure(figsize=(7,5))
plt.bar(glucose_counts.index.astype(str), glucose_counts.values)
plt.title("Glucose Level Distribution")
plt.xlabel("Glucose Level (1=Normal, 2=Above Normal, 3=Well Above Normal)")
plt.ylabel("Number of Patients")
plt.show()
```

**17. How many patients are smokers?**

```python
smokers = (df["smoke"] == 1).sum()
non_smokers = (df["smoke"] == 0).sum()

print("Smokers:", smokers)
print("Non-smokers:", non_smokers)

plt.figure(figsize=(7,5))
plt.bar(["Non-smoker", "Smoker"], [non_smokers, smokers])
plt.title("Smoking Status")
plt.ylabel("Number of Patients")
plt.show()
```

**18. How many patients consume alcohol?**

```python
alcohol_consumers = (df["alco"] == 1).sum()
non_alcohol_consumers = (df["alco"] == 0).sum()

print("Patients who consume alcohol:", alcohol_consumers)
print("Patients who do not consume alcohol:", non_alcohol_consumers)

plt.figure(figsize=(7,5))
plt.bar(["No Alcohol", "Alcohol"], [non_alcohol_consumers, alcohol_consumers])
plt.title("Alcohol Consumption")
plt.ylabel("Number of Patients")
plt.show()
```

**19. How many patients are physically active?**

```python
active_patients = (df["active"] == 1).sum()
inactive_patients = (df["active"] == 0).sum()

print("Physically active patients:", active_patients)
print("Physically inactive patients:", inactive_patients)

plt.figure(figsize=(7,5))
plt.bar(["Inactive", "Active"], [inactive_patients, active_patients])
plt.title("Physical Activity")
plt.ylabel("Number of Patients")
plt.show()
```

**20. Which features have the strongest correlation with cardiovascular disease (cardio)?**

```python
# Calculate correlation of every numeric feature with cardio
correlation_with_cardio = df.corr(numeric_only=True)["cardio"].drop("cardio")
correlation_with_cardio = correlation_with_cardio.sort_values(
    key=lambda x: x.abs(), ascending=False
)

print("Correlation of features with cardio:")
print(correlation_with_cardio)

print("\nStrongest correlations by absolute value:")
print(correlation_with_cardio.head(5))

# Visualize the correlations
plt.figure(figsize=(9,6))
plt.barh(
    correlation_with_cardio.index[::-1],
    correlation_with_cardio.values[::-1]
)
plt.title("Feature Correlation with Cardiovascular Disease (cardio)")
plt.xlabel("Correlation")
plt.ylabel("Feature")
plt.axvline(0, linewidth=1)
plt.show()

strongest_feature = correlation_with_cardio.index[0]
strongest_value = correlation_with_cardio.iloc[0]

print(
    f"\nStrongest feature: {strongest_feature} "
    f"(correlation = {strongest_value:.4f})"
)
```
