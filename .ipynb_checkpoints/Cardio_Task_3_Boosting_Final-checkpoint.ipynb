# Cardiovascular Disease Prediction — EDA + Boosting

## Project Objective
This notebook performs Exploratory Data Analysis (EDA) and builds a **single final ensemble model using Boosting** for cardiovascular disease prediction.

**Initial model accuracy:** 71.36%

**Selected technique:** Gradient Boosting

> This notebook is intentionally structured so that all required libraries are imported before they are used.


# 1. Import All Required Libraries

All libraries used in this notebook are imported in this first code cell.


```python
# Data handling
import os
import numpy as np
import pandas as pd

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Machine Learning
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingClassifier

# Evaluation
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    roc_curve,
    roc_auc_score
)

# Save the trained model
import joblib

print("All required libraries imported successfully.")
```

# 2. Load the Dataset

The notebook expects `cardio_train.csv`.

The original Cardio dataset commonly uses `;` as its separator, so this notebook first tries that format.


```python
# Locate the dataset
possible_paths = [
    "cardio_train.csv",
    "./cardio_train.csv",
    "/mnt/data/cardio_train.csv"
]

csv_path = next((p for p in possible_paths if os.path.exists(p)), None)

if csv_path is None:
    raise FileNotFoundError(
        "cardio_train.csv was not found. "
        "Place cardio_train.csv in the same folder as this notebook and run again."
    )

# Load dataset
try:
    df = pd.read_csv(csv_path, sep=";")

    # Safety check: if the file is comma-separated instead
    if len(df.columns) <= 2:
        df = pd.read_csv(csv_path)
except Exception:
    df = pd.read_csv(csv_path)

print("Dataset loaded successfully.")
print("Dataset shape:", df.shape)

display(df.head())
```

# 3. Basic Dataset Information

```python
print("Column names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nMissing values:")
display(df.isnull().sum().to_frame("Missing Values"))

print("\nDuplicate rows:", df.duplicated().sum())
```

# 4. Descriptive Statistics

```python
display(df.describe().T)
```

# 5. Target Variable — Cardiovascular Disease

The target column is `cardio`.

- `0` = No cardiovascular disease
- `1` = Cardiovascular disease


```python
if "cardio" not in df.columns:
    raise KeyError(
        f"The target column 'cardio' was not found. Available columns: {df.columns.tolist()}"
    )

print("Target distribution:")
display(df["cardio"].value_counts().sort_index())

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x="cardio")
plt.xlabel("Cardio")
plt.ylabel("Count")
plt.title("Cardiovascular Disease Distribution")
plt.show()
```

# 6. Age Distribution

```python
# Convert age from days to years for easier interpretation
if "age" in df.columns:
    df["age_years"] = df["age"] / 365.25

    plt.figure(figsize=(8, 5))
    sns.histplot(df["age_years"], bins=30, kde=True)
    plt.xlabel("Age (years)")
    plt.ylabel("Count")
    plt.title("Age Distribution")
    plt.show()
else:
    print("Age column not found.")
```

# 7. Height and Weight Distribution

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

if "height" in df.columns:
    sns.histplot(df["height"], bins=30, kde=True, ax=axes[0])
    axes[0].set_title("Height Distribution")
    axes[0].set_xlabel("Height")

if "weight" in df.columns:
    sns.histplot(df["weight"], bins=30, kde=True, ax=axes[1])
    axes[1].set_title("Weight Distribution")
    axes[1].set_xlabel("Weight")

plt.tight_layout()
plt.show()
```

# 8. Blood Pressure Analysis

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

if "ap_hi" in df.columns:
    sns.histplot(df["ap_hi"], bins=50, kde=True, ax=axes[0])
    axes[0].set_title("Systolic Blood Pressure (ap_hi)")
    axes[0].set_xlabel("ap_hi")

if "ap_lo" in df.columns:
    sns.histplot(df["ap_lo"], bins=50, kde=True, ax=axes[1])
    axes[1].set_title("Diastolic Blood Pressure (ap_lo)")
    axes[1].set_xlabel("ap_lo")

plt.tight_layout()
plt.show()
```

# 9. Correlation Matrix

```python
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(12, 9))
sns.heatmap(numeric_df.corr(), annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
```

# 10. Data Preparation for Machine Learning

For the final model:

1. `cardio` is used as the target.
2. `id` is removed because it is only an identifier.
3. `age_years` is not used because the original `age` column already contains the same information.
4. Missing rows are removed.
5. No StandardScaler is used because Gradient Boosting uses decision trees and does not require feature scaling.


```python
# Work on a copy so the EDA dataframe remains available
model_df = df.copy()

# Remove the derived age_years column from model features
if "age_years" in model_df.columns:
    model_df = model_df.drop(columns=["age_years"])

# Remove ID column if present
if "id" in model_df.columns:
    model_df = model_df.drop(columns=["id"])

# Remove rows with missing values
model_df = model_df.dropna().copy()

# Separate features and target
X = model_df.drop(columns=["cardio"])
y = model_df["cardio"].astype(int)

# Ensure features are numeric
for column in X.columns:
    if not pd.api.types.is_numeric_dtype(X[column]):
        X[column] = pd.to_numeric(X[column], errors="coerce")

# Drop rows that became missing after numeric conversion
valid_rows = X.notna().all(axis=1) & y.notna()

X = X.loc[valid_rows].copy()
y = y.loc[valid_rows].copy()

print("Features used:")
print(X.columns.tolist())

print("\nFeature shape:", X.shape)
print("Target shape:", y.shape)
```

# 11. Train-Test Split

We use:

- 80% training data
- 20% testing data
- `stratify=y` to maintain the class distribution
- `random_state=42` for reproducible results


```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))
```

# 12. Final Model — Gradient Boosting

### Why Boosting?

The initial model achieved approximately **71.36% accuracy**.

For this project, we selected **Boosting** as the single ensemble technique. Gradient Boosting builds decision trees sequentially, where later trees try to reduce the errors of previous trees.

**No feature scaling is required.**


```python
# Create the Gradient Boosting classifier
model = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)

# Train the model
model.fit(X_train, y_train)

print("Gradient Boosting model trained successfully.")
```

# 13. Predictions and Accuracy

```python
# Predictions
y_pred = model.predict(X_test)

# Probability of class 1
y_prob = model.predict_proba(X_test)[:, 1]

# Accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("Accuracy (%):", accuracy * 100)
```

# 14. Classification Report

```python
print("Classification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        target_names=["No Disease (0)", "Disease (1)"]
    )
)
```

# 15. Confusion Matrix

```python
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)

plt.figure(figsize=(6, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["No Disease", "Disease"],
    yticklabels=["No Disease", "Disease"]
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix — Gradient Boosting")
plt.tight_layout()
plt.show()
```

# 16. ROC Curve and ROC-AUC

```python
auc_score = roc_auc_score(y_test, y_prob)
fpr, tpr, thresholds = roc_curve(y_test, y_prob)

print("ROC-AUC Score:", auc_score)

plt.figure(figsize=(7, 5))
plt.plot(
    fpr,
    tpr,
    label=f"Gradient Boosting (AUC = {auc_score:.3f})"
)
plt.plot([0, 1], [0, 1], linestyle="--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve — Gradient Boosting")
plt.legend()
plt.tight_layout()
plt.show()
```

# 17. Feature Importance

Gradient Boosting provides feature importance values. These show which input features contributed most to the model's decisions.


```python
feature_importance = pd.Series(
    model.feature_importances_,
    index=X.columns
).sort_values(ascending=False)

display(feature_importance.to_frame("Importance"))

plt.figure(figsize=(9, 6))
feature_importance.sort_values().plot(kind="barh")
plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("Feature Importance — Gradient Boosting")
plt.tight_layout()
plt.show()
```

# 18. Final Project Result

```python
print("=" * 50)
print("FINAL MODEL RESULT")
print("=" * 50)
print("Model          : Gradient Boosting")
print(f"Accuracy       : {accuracy * 100:.2f}%")
print(f"ROC-AUC        : {auc_score:.4f}")
print("=" * 50)

print("\nInitial model accuracy : 71.36%")
print(f"Boosting accuracy      : {accuracy * 100:.2f}%")
print(f"Difference             : {(accuracy * 100) - 71.36428571428571:.2f} percentage points")
```

# 19. Save the Final Model

The final Gradient Boosting model is saved as `cardio_boosting_model.pkl`.

A scaler is **not** saved because the final model does not use feature scaling.


```python
model_file = "cardio_boosting_model.pkl"

joblib.dump(model, model_file)

print(f"Model saved successfully as: {model_file}")
```
